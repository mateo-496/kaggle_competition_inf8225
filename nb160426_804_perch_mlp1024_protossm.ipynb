```python
def get_logger(name='birdclef'):
    logger = logging.getLogger(name)
    if not logger.handlers:
        h = logging.StreamHandler()
        h.setFormatter(logging.Formatter('[%(asctime)s] %(levelname)s - %(message)s', '%H:%M:%S'))
        logger.addHandler(h)
        logger.setLevel(logging.INFO)
    return logger

logger = get_logger()

def seed_everything(seed=SEED):
    random.seed(seed); np.random.seed(seed)
    torch.manual_seed(seed); torch.cuda.manual_seed_all(seed)
    os.environ['PYTHONHASHSEED'] = str(seed)

def build_label_map(meta_df, window_df=None):
    species = set(meta_df['primary_label'].unique())
    if window_df is not None:
        for lst in window_df['label_list']:
            species.update(lst)
    species = sorted(species)
    s2i = {s: i for i, s in enumerate(species)}
    return s2i, {i: s for s, i in s2i.items()}

def encode_labels(primary, secondary, s2i):
    vec = np.zeros(len(s2i), dtype=np.float32)
    if primary in s2i: vec[s2i[primary]] = 1.0
    for lbl in (secondary or []):
        if lbl in s2i: vec[s2i[lbl]] = 1.0
    return vec

def competition_score(y_true, y_pred):
    keep = y_true.sum(axis=0) > 0
    return roc_auc_score(y_true[:, keep], y_pred[:, keep], average='macro'), keep

def file_level_auc(y_true, y_pred, n_windows=CLIPS_PER_FILE):
    #assert len(y_true) % n_windows == 0
    n_files = len(y_true) // n_windows
    y_true_f = y_true.reshape(n_files, n_windows, -1).max(axis=1)
    y_pred_f = y_pred.reshape(n_files, n_windows, -1).max(axis=1)
    keep = y_true_f.sum(axis=0) > 0
    if keep.sum() == 0: return 0.0
    return roc_auc_score(y_true_f[:, keep], y_pred_f[:, keep], average='macro')

def mixup_data(x, y, alpha=MIXUP_ALPHA):
    lam = np.random.beta(alpha, alpha) if alpha > 0 else 1.0
    idx = torch.randperm(x.size(0), device=x.device)
    return lam * x + (1 - lam) * x[idx], y, y[idx], lam

def mixup_criterion(criterion, pred, y_a, y_b, lam):
    return lam * criterion(pred, y_a) + (1 - lam) * criterion(pred, y_b)

def save_checkpoint(state, path):
    torch.save(state, path); logger.info(f'Saved → {path}')

@contextmanager
def autocast_ctx():
    if AMP and torch.cuda.is_available():
        with torch.amp.autocast('cuda'): yield
    else: yield

def get_scaler():
    return torch.amp.GradScaler('cuda') if AMP and torch.cuda.is_available() else None

# ── Soundscape label loader ───────────────────────────────────────────────────
def load_soundscape_window_df():
    if not SND_LABELS_CSV.exists():
        logger.warning(f'SND_LABELS_CSV not found: {SND_LABELS_CSV}'); return None
    df = pd.read_csv(SND_LABELS_CSV)
    df.columns = df.columns.str.strip().str.lower()
    if not {'filename', 'primary_label'}.issubset(df.columns):
        logger.warning('Missing required columns'); return None
    if 'end' in df.columns:
        df['end_sec'] = pd.to_timedelta(df['end']).dt.total_seconds().astype(int)
    elif 'end_sec' not in df.columns:
        logger.warning('No end/end_sec column'); return None
    df['stem']       = df['filename'].str.rsplit('.', n=1).str[0]
    df['window_idx'] = (df['end_sec'] // 5 - 1).clip(0, CLIPS_PER_FILE - 1)
    def parse_labels(x):
        if pd.isna(x) or str(x).strip() == '': return []
        return [t.strip() for t in str(x).split(';') if t.strip()]
    df['label_list'] = df['primary_label'].apply(parse_labels)
    agg = (df.groupby(['stem', 'window_idx'])['label_list']
             .apply(lambda lists: sorted(set(sp for lst in lists for sp in lst)))
             .reset_index())
    if HDF5_SND.exists():
        with h5py.File(HDF5_SND, 'r') as f:
            known = set(f.keys())
        agg = agg[agg['stem'].isin(known)]
    logger.info(f'Soundscape windows: {len(agg)}  files: {agg["stem"].nunique()}')
    return agg

```

## Label Mapping

```python
# ── Build label mapping: competition species → Perch bc_index ─────────────────
def build_bc_label_mapping(meta_csv, model_dir):
    """
    Returns:
        BC_INDICES   : np.int32 array of shape (num_classes,)
                       BC_INDICES[i] = Perch label index for train class i,
                       or NO_LABEL_INDEX if unmapped.
        MAPPED_MASK  : bool array (num_classes,)
        MAPPED_POS   : indices of mapped classes in train label space
        MAPPED_BC    : corresponding Perch indices
    """
    meta = pd.read_csv(meta_csv)
    window_df_tmp = load_soundscape_window_df()   
    s2i, i2s = build_label_map(meta, window_df_tmp)
    num_classes = len(s2i)
    PRIMARY_LABELS = [i2s[i] for i in range(num_classes)]

    labels_csv = MODEL_DIR / 'assets/labels.csv'
    bc_labels = (
        pd.read_csv(labels_csv)
        .reset_index()
        .rename(columns={'index': 'bc_index', 'inat2024_fsd50k': 'scientific_name'})
    )
    
    NO_LABEL_INDEX = len(bc_labels)

    # Load taxonomy for scientific name lookup
    taxonomy = pd.read_csv(TAXONOMY_CSV)
    taxonomy['primary_label'] = taxonomy['primary_label'].astype(str)

    bc_lookup = bc_labels.rename(columns={'scientific_name': 'scientific_name_lookup'})
    taxonomy['scientific_name_lookup'] = taxonomy['scientific_name']

    mapping = taxonomy.merge(
        bc_lookup[['scientific_name_lookup', 'bc_index']],
        on='scientific_name_lookup',
        how='left'
    )
    mapping['bc_index'] = mapping['bc_index'].fillna(NO_LABEL_INDEX).astype(int)
    label_to_bc = mapping.set_index('primary_label')['bc_index']

    BC_INDICES = np.array(
        [int(label_to_bc.loc[p]) if p in label_to_bc.index else NO_LABEL_INDEX
         for p in PRIMARY_LABELS],
        dtype=np.int32
    )
    MAPPED_MASK = BC_INDICES != NO_LABEL_INDEX
    MAPPED_POS  = np.where(MAPPED_MASK)[0].astype(np.int32)
    MAPPED_BC   = BC_INDICES[MAPPED_MASK].astype(np.int32)

    n_mapped = MAPPED_MASK.sum()
    print(f'Label mapping: {n_mapped}/{num_classes} train classes mapped to Perch indices')
    return BC_INDICES, MAPPED_MASK, MAPPED_POS, MAPPED_BC, NO_LABEL_INDEX, num_classes
```

## Perch Embedder 

```python
class PerchEmbedder:
    def __init__(self, bc_indices=None, mapped_pos=None, mapped_bc=None, num_classes=None):
        model_path = str(MODEL_DIR)
        print(f'Loading Perch from {model_path}')
        self.model    = tf.saved_model.load(model_path)
        self.infer_fn = self.model.signatures['serving_default']
        # warmup
        _test = tf.constant(np.zeros((1, 160_000), dtype=np.float32))
        _out  = self.infer_fn(inputs=_test)
        self._emb_key = next(k for k in _out if 'embed' in k.lower())
        print(f'emb_key={self._emb_key}  shape={_out[self._emb_key].shape}')
        print('Perch loaded.')

        self.mapped_pos  = mapped_pos   # (num_classes,) indices into train label space
        self.mapped_bc   = mapped_bc    # corresponding Perch indices
        self.num_classes = num_classes

    def embed_and_logits_batch(self, waves: np.ndarray):
        """
        waves: (B, 160_000) float32
        Returns:
            emb    : (B, 1536)
            logits : (B, num_classes) — Perch logits sliced to train classes,
                     zeros for unmapped classes
        """
        out = self.infer_fn(inputs=tf.constant(waves, dtype=tf.float32))
        emb = out['embedding'].numpy().astype(np.float32)
        if emb.ndim > 2:
            emb = emb.reshape(emb.shape[0], -1, emb.shape[-1]).mean(axis=1)

        logits_full = out['label'].numpy().astype(np.float32)  # (B, 14795)

        B = logits_full.shape[0]
        logits_sliced = np.zeros((B, self.num_classes), dtype=np.float32)
        if self.mapped_pos is not None and len(self.mapped_pos) > 0:
            logits_sliced[:, self.mapped_pos] = logits_full[:, self.mapped_bc]

        return emb, logits_sliced

    def embed_batch(self, waves: np.ndarray) -> np.ndarray:
        """Backward-compatible: return embeddings only."""
        emb, _ = self.embed_and_logits_batch(waves)
        return emb
```

# Audio Processing & Augmentations

## Audio Processing

```python
def load_wave(path):
    with warnings.catch_warnings():
        warnings.simplefilter('ignore')
        wave, _ = librosa.load(path, sr=SAMPLE_RATE, mono=True)
    return wave.astype(np.float32)

def pad_or_trim(wave, length=CLIP_SAMPLES):
    if len(wave) < length:
        return np.pad(wave, (0, length - len(wave)))
    start = np.random.randint(0, len(wave) - length + 1)
    return wave[start:start + length]

def center_crop(wave, length=CLIP_SAMPLES):
    if len(wave) <= length:
        return np.pad(wave, (0, length - len(wave)))
    start = (len(wave) - length) // 2
    return wave[start:start + length]

def normalize_wave(wave):
    return wave / (np.abs(wave).max() + 1e-6)

```

## Augmentations

```python
def augment_wave_light(wave):
    """Light augmentation for train_audio clips (2 copies per file)."""
    wave = pad_or_trim(wave)
    if np.random.rand() < 0.8:
        wave = wave + np.random.randn(*wave.shape).astype(np.float32) * np.random.uniform(0.002, 0.015)
    if np.random.rand() < 0.6:
        wave = wave * (10 ** (np.random.uniform(-6, 6) / 20))
    if np.random.rand() < 0.5:
        wave = np.roll(wave, int(np.random.uniform(-0.3, 0.3) * len(wave)))
    return wave.astype(np.float32)

def augment_wave_heavy(wave):
    """Heavy augmentation for soundscape clips (main training data)."""
    wave = pad_or_trim(wave)
    if np.random.rand() < 0.9:
        wave = wave + np.random.randn(*wave.shape).astype(np.float32) * np.random.uniform(0.005, 0.03)
    if np.random.rand() < 0.8:
        wave = wave * (10 ** (np.random.uniform(-9, 9) / 20))
    if np.random.rand() < 0.7:
        wave = np.roll(wave, int(np.random.uniform(-0.5, 0.5) * len(wave)))
    if np.random.rand() < 0.3:
        wave = -wave
    if np.random.rand() < 0.4:
        # mix in a slightly shifted version to simulate room echo
        delay = np.random.randint(100, 800)
        echo  = np.roll(wave, delay) * np.random.uniform(0.05, 0.2)
        wave  = wave + echo
    return wave.astype(np.float32)
```

# Precompute Embeddings

## Precompute Utils

```python
def precompute_train_audio(meta, embedder, augmented=False,
                           n_copies=AUG_COPIES_TRAIN_AUDIO, batch_size=128):
    """Embed train_audio clips → per-sample shards, then merge into HDF5_PATH / HDF5_AUG."""

    shard_dir = HDF5_DIR / ('train_audio_aug_shards' if augmented else 'train_audio_shards')
    shard_dir.mkdir(exist_ok=True)

    # ── Delete stale batch-style shards (batch_XXXXXXX.h5) ───────────────────
    stale = list(shard_dir.glob('batch_*.h5'))
    if stale:
        logger.info(f'Deleting {len(stale)} stale batch shards…')
        for s in stale:
            s.unlink()

    # ── Build work list, skipping already-done per-sample shards ─────────────
    items = []
    for _, row in meta.iterrows():
        copies = n_copies if augmented else 1
        for i in range(copies):
            key  = f'{row["filename"]}__aug{i}' if augmented else row['filename']
            safe = key.replace('/', '_')
            if not (shard_dir / f'{safe}.h5').exists():
                items.append((row['filename'], i, key, augmented))

    done = (len(meta) * (n_copies if augmented else 1)) - len(items)
    print(f'{done} already done, {len(items)} remaining')
    if not items:
        return

    # ── Embed in batches, write one shard per sample ──────────────────────────
    for start in tqdm(range(0, len(items), batch_size),
                      desc=f'train_audio {"augmented" if augmented else "clean"}'):
        batch = items[start:start + batch_size]
        waves, keys = [], []

        for filename, aug_i, key, is_aug in batch:
            try:
                raw = load_wave(TRAIN_AUDIO / filename)
                w = normalize_wave(augment_wave_light(raw) if is_aug else center_crop(raw))
                waves.append(w)
                keys.append(key)
            except Exception as e:
                logger.warning(f'Skip {filename}: {e}')

        if not waves:
            continue

        embs, logits = embedder.embed_and_logits_batch(np.stack(waves))

        for key, emb, lg in zip(keys, embs, logits):
            safe = key.replace('/', '_')
            with h5py.File(shard_dir / f'{safe}.h5', 'w') as f:
                f.attrs['key'] = key
                f.create_dataset('emb',    data=emb, compression='lzf')
                f.create_dataset('logits', data=lg,  compression='lzf')

    logger.info(f'Done. Call merge_train_audio_shards(augmented={augmented}) to consolidate.')


def merge_train_audio_shards(augmented=False):
    """Merge per-sample shards into final HDF5_PATH / HDF5_AUG files."""
    shard_dir    = HDF5_DIR / ('train_audio_aug_shards' if augmented else 'train_audio_shards')
    out_emb_path = HDF5_AUG        if augmented else HDF5_PATH
    out_log_path = HDF5_AUG_LOGITS if augmented else HDF5_LOGITS

    shards = sorted(shard_dir.glob('*.h5'))
    with h5py.File(out_emb_path, 'w') as out_emb, h5py.File(out_log_path, 'w') as out_log:
        for shard in tqdm(shards, desc=f'Merging {"augmented" if augmented else "clean"} train_audio'):
            with h5py.File(shard, 'r') as f:
                key = f.attrs.get('key', shard.stem)
                out_emb.create_dataset(key, data=f['emb'][:],    compression='lzf')
                out_log.create_dataset(key, data=f['logits'][:], compression='lzf')

    logger.info(f'Merged {len(shards)} shards → {out_emb_path} + {out_log_path}')


def precompute_soundscape_embeddings(embedder, augmented=False, n_copies=AUG_COPIES_SOUNDSCAPE, batch_size=16):
    """Embed soundscape clips → per-sample shards, then merge."""
    paths = sorted(TRAIN_SND.glob('*.ogg')) + sorted(TRAIN_SND.glob('*.wav'))
    shard_dir = HDF5_DIR / ('soundscape_aug_shards' if augmented else 'soundscape_shards')
    shard_dir.mkdir(exist_ok=True)

    if augmented:
        items = [
            (p, i, f'{p.stem}__aug{i}')
            for p in paths
            for i in range(n_copies)
            if not (shard_dir / f'{p.stem}__aug{i}.h5').exists()
        ]
    else:
        items = [(p, None, p.stem) for p in paths if not (shard_dir / f'{p.stem}.h5').exists()]

    print(f'{(len(paths) * (n_copies if augmented else 1)) - len(items)} already done, {len(items)} remaining')
    if not items:
        return

    for start in tqdm(range(0, len(items), batch_size),
                      desc=f'soundscape {"augmented" if augmented else "clean"}'):
        batch = items[start:start + batch_size]
        clips_batch, keys = [], []

        for p, aug_i, key in batch:
            wave, _ = sf.read(p, dtype='float32')
            if wave.ndim > 1: wave = wave.mean(axis=1)
            need = CLIPS_PER_FILE * CLIP_SAMPLES
            wave = wave[:need] if len(wave) >= need else np.pad(wave, (0, need - len(wave)))
            clips = wave.reshape(CLIPS_PER_FILE, CLIP_SAMPLES)
            if augmented:
                clips = np.stack([normalize_wave(augment_wave_heavy(c)) for c in clips])
            clips_batch.append(clips)
            keys.append(key)

        clips_flat = np.concatenate(clips_batch, axis=0)
        embs, logits = embedder.embed_and_logits_batch(clips_flat)
        embs   = embs.reshape(len(batch), CLIPS_PER_FILE, PERCH_EMBED_DIM)
        logits = logits.reshape(len(batch), CLIPS_PER_FILE, embedder.num_classes)

        for key, emb, lg in zip(keys, embs, logits):
            with h5py.File(shard_dir / f'{key}.h5', 'w') as f:
                f.attrs['key'] = key
                f.create_dataset('emb',    data=emb, compression='lzf')
                f.create_dataset('logits', data=lg,  compression='lzf')

    logger.info(f'Done. Call merge_soundscape_shards(augmented={augmented}) to consolidate.')


def merge_soundscape_shards(augmented=False):
    shard_dir    = HDF5_DIR / ('soundscape_aug_shards' if augmented else 'soundscape_shards')
    out_emb_path = HDF5_SND_AUG        if augmented else HDF5_SND
    out_log_path = HDF5_SND_AUG_LOGITS if augmented else HDF5_SND_LOGITS

    shards = sorted(shard_dir.glob('*.h5'))
    with h5py.File(out_emb_path, 'w') as out_emb, h5py.File(out_log_path, 'w') as out_log:
        for shard in tqdm(shards, desc=f'Merging {"augmented" if augmented else "clean"} soundscapes'):
            with h5py.File(shard, 'r') as f:
                key = f.attrs.get('key', shard.stem)
                out_emb.create_dataset(key, data=f['emb'][:],    compression='lzf')
                if 'logits' in f:
                    out_log.create_dataset(key, data=f['logits'][:], compression='lzf')

    logger.info(f'Merged {len(shards)} shards → {out_emb_path} + {out_log_path}')
```

# Datasets

## BirdDataset

```python
# ── BirdDataset: flat per-clip embeddings for MLP ─────────────────────────────
class BirdDataset(Dataset):
    """
    Serves flat (1536,) embeddings for MLP training.
    When use_aug=True, interleaves clean embeddings (HDF5_PATH) with augmented
    ones (HDF5_AUG) so every clean sample has AUG_COPIES_TRAIN_AUDIO augmented
    partners in the same epoch.
    """
    def __init__(self, df, s2i, augment=False):
        self.s2i     = s2i
        self.augment = augment
        self._h5     = None
        self._h5_aug = None

        # Build item list: (filename, aug_idx_or_None)
        base = [(row['filename'], None, row['primary_label'],
                 row.get('secondary_labels', None))
                for _, row in df.iterrows()]
        if augment and HDF5_AUG.exists():
            aug_items = [(row['filename'], i, row['primary_label'],
                          row.get('secondary_labels', None))
                         for _, row in df.iterrows()
                         for i in range(AUG_COPIES_TRAIN_AUDIO)]
            self.items = base + aug_items
        else:
            self.items = base

    def _get_h5(self):
        if self._h5 is None: self._h5 = h5py.File(HDF5_PATH, 'r')
        return self._h5

    def _get_h5_aug(self):
        if self._h5_aug is None: self._h5_aug = h5py.File(HDF5_AUG, 'r')
        return self._h5_aug

    def __len__(self): return len(self.items)

    def __getitem__(self, idx):
        filename, aug_i, primary, secondary_raw = self.items[idx]
        if aug_i is None:
            emb = self._get_h5()[filename][:]
        else:
            emb = self._get_h5_aug()[f'{filename}__aug{aug_i}'][:]
        if emb.ndim > 1: emb = emb.reshape(-1, emb.shape[-1]).mean(0)
        label = encode_labels(primary, _parse_secondary(secondary_raw), self.s2i)
        return torch.from_numpy(emb), torch.from_numpy(label)
```

## SoundscapeWindowDataset

```python
# ── SoundscapeWindowDataset: per-5s-window labels for Mamba ──────────────────
class SoundscapeWindowDataset(Dataset):
    """
    Each item → (emb_sequence (12,1536), label_vector) for one 5s window.
    When augment=True and HDF5_SND_AUG exists, aug versions are interleaved.
    Embedding noise augmentation on top of pre-computed augmented embeddings
    is intentionally kept small (the heavy lifting was done at precompute time).
    """
    def __init__(self, window_df, s2i, augment=False):
        self.s2i     = s2i
        self.augment = augment
        self._h5     = None
        self._h5_aug = None

        base = list(window_df.itertuples(index=False))
        if augment and HDF5_SND_AUG.exists():
            aug_rows = []
            for row in base:
                for i in range(AUG_COPIES_SOUNDSCAPE):
                    # Store aug_idx alongside the row by wrapping in a tuple
                    aug_rows.append((row, i))
            self.items     = [(row, None) for row in base] + aug_rows
        else:
            self.items     = [(row, None) for row in base]

    def _get_h5(self):
        if self._h5 is None:     self._h5     = h5py.File(HDF5_SND, 'r')
        return self._h5

    def _get_h5_aug(self):
        if self._h5_aug is None: self._h5_aug = h5py.File(HDF5_SND_AUG, 'r')
        return self._h5_aug

    def _get_h5_logits(self):
        if not hasattr(self, '_h5_logits') or self._h5_logits is None:
            self._h5_logits = h5py.File(HDF5_SND_LOGITS, 'r') if HDF5_SND_LOGITS.exists() else None
        return self._h5_logits

    def __len__(self): return len(self.items)

    def __getitem__(self, idx):
        row, aug_i = self.items[idx]
        stem = row.stem
        if aug_i is None:
            emb = self._get_h5()[stem][:]       # (12, 1536)
        else:
            key = f'{stem}__aug{aug_i}'
            emb = self._get_h5_aug()[key][:]    # (12, 1536)
        # Small embedding-space noise on top (disabled for val)
        if self.augment:
            emb = emb + np.random.randn(*emb.shape).astype(np.float32) * 0.005
        # Load Perch logits for this file (zeros if not available)
        h5l = self._get_h5_logits()
        if h5l is not None and stem in h5l:
            logits = h5l[stem][:]
            if logits.shape[-1] != len(self.s2i):
                # stale precompute — fall back to zeros
                logits = np.zeros((emb.shape[0], len(self.s2i)), dtype=np.float32)
        else:
            logits = np.zeros((emb.shape[0], len(self.s2i)), dtype=np.float32)
        vec = np.zeros(len(self.s2i), dtype=np.float32)
        for sp in row.label_list:
            if sp in self.s2i: vec[self.s2i[sp]] = 1.0
        return torch.from_numpy(emb), torch.from_numpy(logits), torch.from_numpy(vec)
```

## AudioClipSeqDataset

```python
# ── AudioClipSeqDataset: tiled train_audio clips for Mamba ───────────────────
class AudioClipSeqDataset(Dataset):
    """
    Wraps train_audio HDF5 embeddings as (CLIPS_PER_FILE, 1536) sequences
    by tiling short clips.  Serves both clean and augmented versions.
    """
    def __init__(self, df, s2i, augment=False):
        self.s2i     = s2i
        self.augment = augment
        self._h5     = None
        self._h5_aug = None

        base = [(row['filename'], None, row['primary_label'],
                 row.get('secondary_labels', None))
                for _, row in df.iterrows()]
        if augment and HDF5_AUG.exists():
            aug_items = [(row['filename'], i, row['primary_label'],
                          row.get('secondary_labels', None))
                         for _, row in df.iterrows()
                         for i in range(AUG_COPIES_TRAIN_AUDIO)]
            self.items = base + aug_items
        else:
            self.items = base

    def _get_h5(self):
        if self._h5 is None: self._h5 = h5py.File(HDF5_PATH, 'r')
        return self._h5

    def _get_h5_aug(self):
        if self._h5_aug is None: self._h5_aug = h5py.File(HDF5_AUG, 'r')
        return self._h5_aug

    def __len__(self): return len(self.items)

    def __getitem__(self, idx):
        filename, aug_i, primary, secondary_raw = self.items[idx]
        if aug_i is None:
            emb = self._get_h5()[filename][:]
        else:
            emb = self._get_h5_aug()[f'{filename}__aug{aug_i}'][:]
        if emb.ndim == 3: emb = emb.mean(axis=1)
        if emb.ndim == 1: emb = emb[np.newaxis, :]
        T = emb.shape[0]
        if T < CLIPS_PER_FILE:
            emb = np.tile(emb, (-(-CLIPS_PER_FILE // T), 1))[:CLIPS_PER_FILE]
        else:
            emb = emb[:CLIPS_PER_FILE]
        if self.augment:
            emb = np.roll(emb, np.random.randint(0, CLIPS_PER_FILE), axis=0)
            emb = emb + np.random.randn(*emb.shape).astype(np.float32) * 0.005
        label = encode_labels(primary, _parse_secondary(secondary_raw), self.s2i)
        perch_logits = np.zeros((CLIPS_PER_FILE, len(self.s2i)), dtype=np.float32)
        return torch.from_numpy(emb), torch.from_numpy(perch_logits), torch.from_numpy(label)
```

## Dataset Utils

```python
def _parse_secondary(raw):
    if pd.isna(raw) or not isinstance(raw, str) or raw.strip() in ('', '[]'):
        return []
    return [s.strip().strip("'\"") for s in raw.strip('[]').split(',') if s.strip()]

# ── Fold split helpers ────────────────────────────────────────────────────────
def get_fold_dfs(meta, fold=FOLD, n_folds=NUM_FOLDS, seed=SEED):
    skf = StratifiedKFold(n_splits=n_folds, shuffle=True, random_state=seed)
    for f, (tr, val) in enumerate(skf.split(meta, meta['primary_label'])):
        if f == fold: return meta.iloc[tr].copy(), meta.iloc[val].copy()
    raise ValueError(f'Fold {fold} not found')


def get_loaders(meta, s2i, fold=FOLD):
    """MLP loaders — flat per-clip embeddings, train_audio only."""    
    train_df, val_df = get_fold_dfs(meta, fold=fold)
    nw = NUM_WORKERS
    train_loader = DataLoader(
        BirdDataset(train_df, s2i, augment=True),
        batch_size=BATCH_SIZE, shuffle=True, num_workers=nw,
        pin_memory=PIN_MEMORY, drop_last=True, persistent_workers=nw > 0,
    )
    val_loader = DataLoader(
        BirdDataset(val_df, s2i, augment=False),
        batch_size=BATCH_SIZE * 2, shuffle=False, num_workers=nw,
        pin_memory=PIN_MEMORY, drop_last=False, persistent_workers=nw > 0,
    )
    return train_loader, val_loader


def get_proto_loaders(meta, s2i, fold=FOLD):
    """
    Mamba train: soundscape per-window sequences (primary, with augmented copies) +
                 tiled train_audio clips (secondary).
    Val: soundscape held-out files only, no augmentation, file-major order.
    """
    window_df = load_soundscape_window_df()
    snd_train_dataset = snd_val_dataset = None

    if window_df is not None and len(window_df) > 0:
        gkf    = GroupKFold(n_splits=NUM_FOLDS)
        dummy  = np.zeros(len(window_df))
        splits = list(gkf.split(dummy, groups=window_df['stem'].values))
        tr_idx, val_idx = splits[fold]

        snd_train_df = window_df.iloc[tr_idx].reset_index(drop=True)
        snd_val_df   = window_df.iloc[val_idx].reset_index(drop=True)

        val_counts = snd_val_df.groupby('stem')['window_idx'].count()
        full_stems = val_counts[val_counts == CLIPS_PER_FILE].index
        snd_val_df = (snd_val_df
              .sort_values(['stem', 'window_idx'])
              .reset_index(drop=True))

        snd_train_dataset = SoundscapeWindowDataset(snd_train_df, s2i, augment=True)
        snd_val_dataset   = SoundscapeWindowDataset(snd_val_df,   s2i, augment=False)
        logger.info(f'Soundscape train: {len(snd_train_dataset)}  val: {len(snd_val_dataset)}  val_files: {len(full_stems)}')
    else:
        logger.warning('No soundscape data — Proto trains on audio clips only.')

    train_df, val_df    = get_fold_dfs(meta, fold=fold)
    audio_train_dataset = AudioClipSeqDataset(train_df, s2i, augment=True)

    train_dataset = (torch.utils.data.ConcatDataset([snd_train_dataset, audio_train_dataset])
                     if snd_train_dataset else audio_train_dataset)

    if snd_val_dataset:
        val_dataset      = snd_val_dataset
        val_is_soundscape = True
    else:
        val_dataset      = AudioClipSeqDataset(val_df, s2i, augment=False)
        val_is_soundscape = False

    nw = NUM_WORKERS
    train_loader = DataLoader(train_dataset, batch_size=PROTO_BATCH, shuffle=True,
                              num_workers=nw, pin_memory=PIN_MEMORY, drop_last=True,
                              persistent_workers=nw > 0)
    val_loader   = DataLoader(val_dataset,   batch_size=PROTO_BATCH * 2, shuffle=False,
                              num_workers=nw, pin_memory=PIN_MEMORY, drop_last=False,
                              persistent_workers=nw > 0)
    logger.info(f'Proto train: {len(train_dataset)}  val: {len(val_dataset)}  val_file_level={val_is_soundscape}')
    return train_loader, val_loader, val_is_soundscape
```

# Models

## MLP Head

```python
# ─── MLP Head (unchanged) ─────────────────────────────────────────────────────
class PerchMLP(nn.Module):
    def __init__(self, num_classes, embed_dim=PERCH_EMBED_DIM):
        super().__init__()
        self.net = nn.Sequential(
            nn.LayerNorm(embed_dim),
            nn.Linear(embed_dim, 1024),
            nn.GELU(),
            nn.Dropout(DROP_RATE),
            nn.Linear(1024, num_classes),
        )
    def forward(self, x): return self.net(x)
```

## SelectiveSSM

```python
class SelectiveSSM(nn.Module):
    def __init__(self, d_model, d_state=16, d_conv=4):
        super().__init__()
        self.d_model = d_model
        self.d_state = d_state
        self.in_proj  = nn.Linear(d_model, 2 * d_model, bias=False)
        self.conv1d   = nn.Conv1d(d_model, d_model, d_conv, padding=d_conv - 1, groups=d_model)
        self.dt_proj  = nn.Linear(d_model, d_model, bias=True)
        A = torch.arange(1, d_state + 1, dtype=torch.float32).unsqueeze(0).expand(d_model, -1)
        self.A_log = nn.Parameter(torch.log(A))
        self.D     = nn.Parameter(torch.ones(d_model))
        self.B_proj = nn.Linear(d_model, d_state, bias=False)
        self.C_proj = nn.Linear(d_model, d_state, bias=False)
        self.out_proj = nn.Linear(d_model, d_model, bias=False)

    def forward(self, x):
        B_size, T, D = x.shape
        xz = self.in_proj(x)
        x_ssm, z = xz.chunk(2, dim=-1)
        x_conv = self.conv1d(x_ssm.transpose(1, 2))[:, :, :T].transpose(1, 2)
        x_conv = F.silu(x_conv)
        dt = F.softplus(self.dt_proj(x_conv))
        A  = -torch.exp(self.A_log)
        B  = self.B_proj(x_conv)
        C  = self.C_proj(x_conv)
        h = torch.zeros(B_size, D, self.d_state, device=x.device)
        ys = []
        for t in range(T):
            dA = torch.exp(A[None, :, :] * dt[:, t, :, None])
            dB = dt[:, t, :, None] * B[:, t, None, :]
            h  = h * dA + x[:, t, :, None] * dB
            ys.append((h * C[:, t, None, :]).sum(-1))
        y = torch.stack(ys, dim=1)
        return y + x * self.D[None, None, :]

```

## Temporal Cross-Attention

```python
class TemporalCrossAttention(nn.Module):
    def __init__(self, d_model, n_heads=4, dropout=0.1):
        super().__init__()
        self.attn  = nn.MultiheadAttention(d_model, n_heads, dropout=dropout, batch_first=True)
        self.norm  = nn.LayerNorm(d_model)
        self.ffn   = nn.Sequential(
            nn.Linear(d_model, d_model * 2), 
            nn.GELU(), 
            nn.Dropout(dropout),
            nn.Linear(d_model * 2, d_model), 
            nn.Dropout(dropout),
        )
        self.norm2 = nn.LayerNorm(d_model)

    def forward(self, x):
        residual = x
        x = self.norm(x)
        attn_out, _ = self.attn(x, x, x)
        x = residual + attn_out
        residual = x
        x = self.norm2(x)
        return residual + self.ffn(x)
```

## ProtoSSM

```python
class ProtoSSM(nn.Module):
    """
    Processes a (B, T, 1536) embedding sequence and fuses with (B, T, num_classes)
    Perch logits via a per-class learnable gate (fusion_alpha).

    Forward signature:
        emb          : (B, T, d_input)
        perch_logits : (B, T, num_classes)  — can be None
        site_ids     : (B,) long
        hours        : (B,) long
    Returns:
        species_logits : (B, T, num_classes)
        family_logits  : (B, n_families) or None
        h_temporal     : (B, T, d_model)
    """
    def __init__(self, d_input=1536, d_model=256, d_state=16,
                 n_ssm_layers=3, n_classes=206, n_windows=12,
                 dropout=0.15, n_sites=20, meta_dim=16,
                 use_cross_attn=True, cross_attn_heads=4):
        super().__init__()
        self.d_model   = d_model
        self.n_classes = n_classes
        self.n_windows = n_windows

        self.input_proj = nn.Sequential(
            nn.Linear(d_input, d_model), nn.LayerNorm(d_model),
            nn.GELU(), nn.Dropout(dropout),
        )
        self.pos_enc = nn.Parameter(torch.randn(1, n_windows, d_model) * 0.02)

        self.site_emb  = nn.Embedding(n_sites, meta_dim)
        self.hour_emb  = nn.Embedding(24, meta_dim)
        self.meta_proj = nn.Linear(2 * meta_dim, d_model)

        self.ssm_fwd   = nn.ModuleList([SelectiveSSM(d_model, d_state) for _ in range(n_ssm_layers)])
        self.ssm_bwd   = nn.ModuleList([SelectiveSSM(d_model, d_state) for _ in range(n_ssm_layers)])
        self.ssm_merge = nn.ModuleList([nn.Linear(2 * d_model, d_model) for _ in range(n_ssm_layers)])
        self.ssm_norm  = nn.ModuleList([nn.LayerNorm(d_model)            for _ in range(n_ssm_layers)])
        self.ssm_drop  = nn.Dropout(dropout)

        self.use_cross_attn = use_cross_attn
        if use_cross_attn:
            self.cross_attn = TemporalCrossAttention(d_model, n_heads=cross_attn_heads, dropout=dropout)

        self.prototypes = nn.Parameter(torch.randn(n_classes, d_model) * 0.02)
        self.proto_temp = nn.Parameter(torch.tensor(5.0))
        self.class_bias = nn.Parameter(torch.zeros(n_classes))

        # Per-class learnable gate: alpha * sim + (1-alpha) * perch_logit
        self.fusion_alpha = nn.Parameter(torch.zeros(n_classes))

        self.n_families  = 0
        self.family_head = None

    def init_prototypes_from_data(self, embeddings, labels):
        with torch.no_grad():
            h = self.input_proj(embeddings)
            for c in range(self.n_classes):
                mask = labels[:, c] > 0.5
                if mask.sum() > 0:
                    self.prototypes.data[c] = F.normalize(h[mask].mean(0), dim=0)

    def init_family_head(self, n_families, class_to_family):
        self.n_families  = n_families
        self.family_head = nn.Linear(self.d_model, n_families)
        self.register_buffer('class_to_family',
                              torch.tensor(class_to_family, dtype=torch.long))

    def forward(self, emb, perch_logits=None, site_ids=None, hours=None):
        B, T, _ = emb.shape
        h = self.input_proj(emb) + self.pos_enc[:, :T, :]

        if site_ids is not None and hours is not None:
            s_e  = self.site_emb(site_ids.clamp(0, self.site_emb.num_embeddings - 1))
            h_e  = self.hour_emb(hours.clamp(0, 23))
            meta = self.meta_proj(torch.cat([s_e, h_e], dim=-1))
            h    = h + meta[:, None, :]

        for fwd, bwd, merge, norm in zip(self.ssm_fwd, self.ssm_bwd, self.ssm_merge, self.ssm_norm):
            residual = h
            h_f = fwd(h)
            h_b = bwd(h.flip(1)).flip(1)
            h   = merge(torch.cat([h_f, h_b], dim=-1))
            h   = self.ssm_drop(h)
            h   = norm(h + residual)

        if self.use_cross_attn:
            h = self.cross_attn(h)

        h_temporal = h
        h_norm  = F.normalize(h, dim=-1)
        p_norm  = F.normalize(self.prototypes, dim=-1)
        temp    = F.softplus(self.proto_temp)
        sim     = torch.matmul(h_norm, p_norm.T) * temp + self.class_bias[None, None, :]

        if perch_logits is not None:
            alpha          = torch.sigmoid(self.fusion_alpha)[None, None, :]
            species_logits = alpha * sim + (1 - alpha) * perch_logits
        else:
            species_logits = sim

        family_logits = None
        if self.family_head is not None:
            family_logits = self.family_head(h.mean(dim=1))

        return species_logits, family_logits, h_temporal

    def count_parameters(self):
        return sum(p.numel() for p in self.parameters() if p.requires_grad)
```

# Training

## Training Utils

```python
class BCEWithLabelSmoothing(nn.Module):
    def __init__(self, pos_weight, smoothing=LABEL_SMOOTHING):
        super().__init__()
        self.smoothing = smoothing
        self.bce = nn.BCEWithLogitsLoss(pos_weight=pos_weight)
    def forward(self, logits, targets):
        return self.bce(logits, targets * (1 - self.smoothing) + self.smoothing / 2)


def get_pos_weight(train_df, s2i, device):
    counts = np.zeros(len(s2i), dtype=np.float32)
    for _, row in train_df.iterrows():
        if row['primary_label'] in s2i:
            counts[s2i[row['primary_label']]] += 1
    pw = np.clip((len(train_df) - counts) / (counts + 1e-6), 1.0, 100.0)
    return torch.from_numpy(pw).to(device)


def mixup_seq(x, y, alpha=MIXUP_ALPHA):
    lam = np.random.beta(alpha, alpha) if alpha > 0 else 1.0
    idx = torch.randperm(x.size(0), device=x.device)
    return lam * x + (1 - lam) * x[idx], y, y[idx], lam


def focal_bce(logits, targets, gamma=FOCAL_GAMMA, pos_weight=None):
    """Focal BCE loss for multi-label classification."""
    if pos_weight is not None:
        bce = F.binary_cross_entropy_with_logits(
            logits, targets, pos_weight=pos_weight, reduction='none')
    else:
        bce = F.binary_cross_entropy_with_logits(logits, targets, reduction='none')
    p  = torch.sigmoid(logits)
    pt = targets * p + (1 - targets) * (1 - p)
    return ((1 - pt) ** gamma * bce).mean()
```

## One Epoch Training

```python
def train_proto_epoch(model, loader, optimizer, scheduler, device,
                      pos_weight=None, distill_w=0.1):
    """Train one epoch of ProtoSSM with (emb, logits, labels) batches."""
    model.train()
    total_loss = 0.0
    for embs, perch_logits, labels in tqdm(loader, desc='  proto-train', leave=False):
        embs        = embs.to(device)        # (B, 12, 1536)
        perch_logits = perch_logits.to(device)  # (B, 12, num_classes)
        labels      = labels.to(device)     # (B, num_classes)
        # Broadcast labels over time axis
        labels_seq = labels.unsqueeze(1).expand(-1, embs.shape[1], -1)

        # Mixup at file level
        if MIXUP_ALPHA > 0:
            lam = float(np.random.beta(MIXUP_ALPHA, MIXUP_ALPHA))
            lam = max(lam, 1 - lam)
            perm = torch.randperm(embs.size(0), device=device)
            embs         = lam * embs        + (1 - lam) * embs[perm]
            perch_logits = lam * perch_logits + (1 - lam) * perch_logits[perm]
            labels_seq   = lam * labels_seq  + (1 - lam) * labels_seq[perm]

        species_out, _, _ = model(embs, perch_logits)

        pw = pos_weight[None, None, :] if pos_weight is not None else None
        loss_main    = focal_bce(species_out, labels_seq, pos_weight=pw)
        loss_distill = F.mse_loss(species_out, perch_logits)
        loss = loss_main + distill_w * loss_distill

        optimizer.zero_grad()
        loss.backward()
        nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        if scheduler is not None:
            scheduler.step()
        total_loss += loss.item()
    return total_loss / max(len(loader), 1)

def train_one_epoch(model, loader, criterion, optimizer, scheduler, scaler, device):
    model.train()
    total_loss = 0.0
    optimizer.zero_grad()
    for step, (embs, labels) in enumerate(tqdm(loader, desc='  train', leave=False)):
        embs, labels = embs.to(device), labels.to(device)
        embs, y_a, y_b, lam = mixup_seq(embs, labels)
        with autocast_ctx():
            loss = mixup_criterion(criterion, model(embs), y_a, y_b, lam) / ACCUM_STEPS
        (scaler.scale(loss) if scaler else loss).backward()
        if (step + 1) % ACCUM_STEPS == 0:
            if scaler:
                scaler.unscale_(optimizer)
                nn.utils.clip_grad_norm_(model.parameters(), 1.0)
                scaler.step(optimizer); scaler.update()
            else:
                nn.utils.clip_grad_norm_(model.parameters(), 1.0)
                optimizer.step()
            optimizer.zero_grad()
            scheduler.step()
        total_loss += loss.item() * ACCUM_STEPS
    return total_loss / len(loader)
```

## Validation

```python
@torch.no_grad()
def validate_proto(model, loader, device, distill_w=0.1):
    """Validate ProtoSSM; returns (loss, auc)."""
    model.eval()
    all_preds, all_labels = [], []
    total_loss = 0.0
    for embs, perch_logits, labels in tqdm(loader, desc='  proto-val', leave=False):
        embs         = embs.to(device)
        perch_logits = perch_logits.to(device)
        labels       = labels.to(device)
        labels_seq   = labels.unsqueeze(1).expand(-1, embs.shape[1], -1)
        species_out, _, _ = model(embs, perch_logits)
        loss_main    = F.binary_cross_entropy_with_logits(species_out, labels_seq)
        loss_distill = F.mse_loss(species_out, perch_logits)
        total_loss   += (loss_main + distill_w * loss_distill).item()
        # Use last-window prediction for AUC
        all_preds.append(torch.sigmoid(species_out[:, -1, :]).cpu().numpy())
        all_labels.append(labels.cpu().numpy())
    all_preds  = np.concatenate(all_preds)
    all_labels = np.concatenate(all_labels)
    try:
        keep = all_labels.sum(axis=0)>0
        auc = roc_auc_score(all_labels[:keep], all_preds[:keep], average='macro') if keep.sum() >= 2 else float('nan')
    except Exception:
        auc = float('nan')
    return total_loss / max(len(loader), 1), auc


@torch.no_grad()
def validate(model, loader, criterion, device, file_level=False):
    model.eval()
    total_loss, all_preds, all_labels = 0.0, [], []
    for embs, labels in tqdm(loader, desc='  valid', leave=False):
        embs, labels = embs.to(device), labels.to(device)
        with autocast_ctx():
            logits = model(embs)
            total_loss += criterion(logits, labels).item()
        all_preds.append(torch.sigmoid(logits).cpu().numpy())
        all_labels.append(labels.cpu().numpy())
    all_preds  = np.concatenate(all_preds)
    all_labels = np.concatenate(all_labels)
    if file_level and len(all_preds) % CLIPS_PER_FILE == 0:
        roc_auc = file_level_auc(all_labels, all_preds)
    else:
        roc_auc, _ = competition_score(all_labels, all_preds)
    return total_loss / len(loader), roc_auc
```

## Complete Training

```python
def train_mlp(model, s2i, epochs=TRAIN_EPOCHS):
    seed_everything()
    meta        = pd.read_csv(META_CSV)
    train_loader, val_loader = get_loaders(meta, s2i, fold=FOLD)
    train_df, _ = get_fold_dfs(meta)
    criterion   = BCEWithLabelSmoothing(get_pos_weight(train_df, s2i, DEVICE))
    optimizer   = AdamW(model.parameters(), lr=LR, weight_decay=WEIGHT_DECAY)
    scheduler   = OneCycleLR(optimizer, max_lr=LR,
                              total_steps=epochs * len(train_loader),
                              pct_start=WARMUP_EPOCHS / epochs, anneal_strategy='cos')
    scaler    = get_scaler()
    best_score = 0.0
    ckpt_path  = CKPT_DIR / f'fold{FOLD}_best_mlp.pt'
    for epoch in range(1, epochs + 1):
        logger.info(f'[mlp] Epoch {epoch}/{epochs}')
        tr_loss = train_one_epoch(model, train_loader, criterion, optimizer, scheduler, scaler, DEVICE)
        val_loss, roc_auc = validate(model, val_loader, criterion, DEVICE)
        logger.info(f'  tr={tr_loss:.4f}  val={val_loss:.4f}  auc={roc_auc:.4f}')
        if roc_auc > best_score:
            best_score = roc_auc
            save_checkpoint({'epoch': epoch, 'model': model.state_dict(), 'best_score': best_score}, ckpt_path)
    logger.info(f'[mlp] Best auc: {best_score:.4f}')
    return ckpt_path

def train_proto(model, s2i, epochs=TRAIN_EPOCHS):
    seed_everything()
    meta          = pd.read_csv(META_CSV)
    window_df_tmp = load_soundscape_window_df()
    s2i, _        = build_label_map(meta, window_df_tmp)
    train_loader, val_loader, _ = get_proto_loaders(meta, s2i, fold=FOLD)
    train_df, _   = get_fold_dfs(meta)
    pos_weight    = get_pos_weight(train_df, s2i, DEVICE)
    optimizer     = AdamW(model.parameters(), lr=LR, weight_decay=WEIGHT_DECAY)
    scheduler     = OneCycleLR(optimizer, max_lr=LR,
                              total_steps=epochs * len(train_loader),
                              pct_start=WARMUP_EPOCHS / epochs, anneal_strategy='cos')
    best_loss = float('inf')
    ckpt_path  = CKPT_DIR / f'fold{FOLD}_best_proto.pt'
    for epoch in range(1, epochs + 1):
        logger.info(f'[proto] Epoch {epoch}/{epochs}')
        tr_loss = train_proto_epoch(model, train_loader, optimizer, scheduler, DEVICE,
                                    pos_weight=pos_weight)
        val_loss, roc_auc = validate_proto(model, val_loader, DEVICE)
        logger.info(f'  tr={tr_loss:.4f}  val={val_loss:.4f} ')
        if val_loss < best_loss:
            best_loss = val_loss
            save_checkpoint({'epoch': epoch, 'model': model.state_dict(), 'best_loss': best_loss}, ckpt_path)
    logger.info(f'[proto] Best val_loss: {best_loss:.4f}')
    return ckpt_path
```

# Ensemble Weight Calibration

```python
def sweep_ensemble_weights_on_soundscapes(mlp, proto, s2i):
    """
    Run both models on ALL labeled train_soundscapes and sweep blend weight.
    Uses SND_LABELS_CSV as ground truth — same data the competition scores on.
    """
    window_df = load_soundscape_window_df()
    if window_df is None or len(window_df) == 0:
        logger.warning('No soundscape labels found — defaulting to 0.5/0.5')
        return np.array([0.5, 0.5], dtype=np.float32)

    # Full soundscape dataset (no fold split, no augmentation)
    full_dataset = SoundscapeWindowDataset(window_df, s2i, augment=False)
    full_loader  = DataLoader(full_dataset, batch_size=PROTO_BATCH * 2,
                              shuffle=False, num_workers=NUM_WORKERS,
                              pin_memory=PIN_MEMORY)

    mlp.eval(); proto.eval()
    p_mlp, p_proto, y_true = [], [], []

    with torch.no_grad():
        for embs, perch_logits, labels in tqdm(full_loader, desc='Ensemble sweep'):
            # MLP: mean-pool sequence → flat embedding
            flat = embs.mean(dim=1).to(DEVICE)
            p_mlp.append(torch.sigmoid(mlp(flat)).cpu().numpy())

            # Proto: takes full sequence
            out, _, _ = proto(embs.to(DEVICE), perch_logits.to(DEVICE))
            p_proto.append(torch.sigmoid(out[:, -1, :]).cpu().numpy())

            y_true.append(labels.numpy())

    p_mlp   = np.concatenate(p_mlp)
    p_proto = np.concatenate(p_proto)
    y_true  = np.concatenate(y_true)

    keep = y_true.sum(axis=0) > 0

    best_auc, best_w = 0.0, 0.5
    for w in np.arange(0.0, 1.05, 0.05):
        blend = w * p_mlp + (1 - w) * p_proto
        try:
            auc = roc_auc_score(y_true[:, keep], blend[:, keep], average='macro')
        except Exception:
            continue
        if auc > best_auc:
            best_auc, best_w = auc, w

    logger.info(f'MLP-only  AUC={roc_auc_score(y_true[:,keep], p_mlp[:,keep],   average="macro"):.4f}')
    logger.info(f'Proto-only AUC={roc_auc_score(y_true[:,keep], p_proto[:,keep], average="macro"):.4f}')
    logger.info(f'Best blend w_mlp={best_w:.2f}  w_proto={1-best_w:.2f}  AUC={best_auc:.4f}')

    return np.array([best_w, 1.0 - best_w], dtype=np.float32)


if TRAIN_MODE:
    weights = sweep_ensemble_weights_on_soundscapes(mlp, proto, s2i)
    np.save(ENSEMBLE_W, weights)
    print(f'Saved weights: {weights}  → {ENSEMBLE_W}')
```

# Inference

## Inference Utils

```python
def sigmoid(x): return 1.0 / (1.0 + np.exp(-x))

def predict_ensemble(embs_flat: np.ndarray, logits_flat: np.ndarray) -> np.ndarray:
    """
    embs_flat   : (B, 1536) — B = batch_n * CLIPS_PER_FILE, file-major order.
    logits_flat : (B, num_classes) — Perch logits sliced to train classes.
    returns     : (B, num_classes) probabilities
    """
    # MLP: operates on flat embeddings (mean-pooled per window already 1-D)
    logits_mlp = sess_mlp.run(['logits'], {'embedding': embs_flat})[0]
    p_mlp      = sigmoid(logits_mlp)

    B = embs_flat.shape[0]
    assert B % CLIPS_PER_FILE == 0
    embs_seq    = embs_flat.reshape(B // CLIPS_PER_FILE, CLIPS_PER_FILE, PERCH_EMBED_DIM)
    logits_seq  = logits_flat.reshape(B // CLIPS_PER_FILE, CLIPS_PER_FILE, -1)

    # ProtoSSM: takes (embs_seq, perch_logits_seq), returns last-window logit per file
    logits_proto = sess_proto.run(
        ['logits'],
        {'embedding': embs_seq, 'perch_logits': logits_seq}
    )[0]                                                    # (n_files, num_classes)
    p_proto = np.repeat(sigmoid(logits_proto), CLIPS_PER_FILE, axis=0)

    return w_mlp * p_mlp + w_proto * p_proto

```

## Inference Exec

```python
train_species = [i2s[i] for i in range(num_classes)]
sub_col_idx   = {sp: i for i, sp in enumerate(sub_species)}
train_to_sub  = {ti: sub_col_idx[sp] for ti, sp in enumerate(train_species) if sp in sub_col_idx}
uniform_prior = 1.0 / len(sub_species)
logger.info(f'Mapped: {len(train_to_sub)}/{num_classes}  prior={uniform_prior:.4f}')

embedder = PerchEmbedder(bc_indices=BC_INDICES, mapped_pos=MAPPED_POS, mapped_bc=MAPPED_BC, num_classes=num_classes)

stems = ss['row_id'].str.rsplit('_', n=1).str[0].unique()
soundscape_paths = []
for stem in stems:
    for ext in ('.ogg', '.wav', '.flac'):
        p = TEST_SND_DIR / (stem + ext)
        if p.exists(): soundscape_paths.append(p); break
if not soundscape_paths:
    soundscape_paths = sorted(TEST_SND_DIR.glob('*.ogg')) + sorted(TEST_SND_DIR.glob('*.wav'))
logger.info(f'Soundscapes: {len(soundscape_paths)}')

BATCH_FILES = 16
rows = []
t_load = t_embed = t_infer = 0.0

for start in tqdm(range(0, len(soundscape_paths), BATCH_FILES), desc='Inference'):
    batch_paths = soundscape_paths[start:start + BATCH_FILES]
    batch_n = len(batch_paths)

    t0 = time.perf_counter()
    x = np.zeros((batch_n * CLIPS_PER_FILE, CLIP_SAMPLES), dtype=np.float32)
    meta_batch = []
    for i, path in enumerate(batch_paths):
        wave, _ = sf.read(path, dtype='float32')
        if wave.ndim > 1: wave = wave.mean(axis=1)
        need = CLIPS_PER_FILE * CLIP_SAMPLES
        wave = wave[:need] if len(wave) >= need else np.pad(wave, (0, need - len(wave)))
        x[i*CLIPS_PER_FILE:(i+1)*CLIPS_PER_FILE] = wave.reshape(CLIPS_PER_FILE, CLIP_SAMPLES)
        meta_batch.extend([(path.stem, t) for t in range(5, 65, 5)])
    t_load += time.perf_counter() - t0

    t0 = time.perf_counter()
    all_embs, all_logits = embedder.embed_and_logits_batch(x)
    t_embed += time.perf_counter() - t0

    t0 = time.perf_counter()
    all_probs = predict_ensemble(all_embs, all_logits)
    t_infer += time.perf_counter() - t0

    for (stem, end_t), probs in zip(meta_batch, all_probs):
        out = np.full(len(sub_species), uniform_prior, dtype=np.float32)
        for ti, si in train_to_sub.items(): out[si] = probs[ti]
        rows.append({'row_id': f'{stem}_{end_t}', **dict(zip(sub_species, out.tolist()))})

print(f'Total — load={t_load:.2f}s  embed={t_embed:.2f}s  infer={t_infer:.2f}s')

if not rows:
    logger.warning('No rows — filling with uniform prior')
    sub_final = ss.copy()
    for sp in sub_species: sub_final[sp] = uniform_prior
else:
    sub_pred  = pd.DataFrame(rows)
    sub_final = ss[['row_id']].merge(sub_pred, on='row_id', how='left')
    for sp in sub_species: sub_final[sp] = sub_final[sp].fillna(uniform_prior)

sub_final.to_csv(OUT_PATH, index=False)
logger.info(f'Saved → {OUT_PATH}  shape={sub_final.shape}')
sub_final.head(3)
```
